```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import joblib

# Veriyi yükle
df = pd.read_csv('personel_ai_veri_seti.csv')

# Kategorik verileri sayıya çevir (Modelin anlaması için)
# Not: Eğitimde hangi dönüşümleri yaptıysan aynısını yapmalısın.
# Şimdilik en temel haliyle eğitiyoruz:
X = df[['Yas', 'Boy', 'Kilo', 'VKI', 'Sakatlik_Siddeti', 'Spor_Gecmisi_Yil']]
y = df['Program_Etiketi']

# Modeli eğit
model = RandomForestClassifier(n_estimators=100)
model.fit(X, y)

# MODELİ KAYDET (İşte eksik olan dosya burada oluşuyor)
joblib.dump(model, 'fitness_model.pkl')
print("Model başarıyla 'fitness_model.pkl' adıyla kaydedildi!")
```
